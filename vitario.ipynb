# ViTario

ViTario is an AI system that uses a custom vision transformer (ViT) model to predict the next move to make at a given frame in the Super Mario Bros. game on the Nintendo Entertainment System (NES).

Check it out below!

#### Training and Evaluation

```python
from vitario import ViTario
```

```python
# train ViTario model
v = ViTario()
v.train(batch_size=64, epochs=10, test_size=0.2)
```

#### Example Predictions

```python
from PIL import Image
from vitario import get_next_moves_from_ints
```

```python
# get first sample of frame and expected move
sample_frame1 = Image.open(
    './data-smb/Rafael_zzmwgrxt_e1_8-1_win/Rafael_zzmwgrxt_e1_8-1_f274_a20_2019-04-14_03-17-22.win.png').convert('RGB')
expected_move1 = get_next_moves_from_ints([20])[0]
```

```python
# get second sample of frame and expected move
sample_frame2 = Image.open(
    'data-smb/Rafael_zzmwgrxt_e1_8-1_win/Rafael_zzmwgrxt_e1_8-1_f3477_a148_2019-04-14_03-18-27.win.png').convert('RGB')
expected_move2 = get_next_moves_from_ints([148])[0]
```

```python
# predict next moves
next_moves = v.predict_next_move([sample_frame1, sample_frame2])
```

```python
# show frame
sample_frame1
```

```python
# show expected and actual moves
print(f'Expected next move: {expected_move1}')
print(
    f'Predicted next move: {next_moves[0][0]}. Probability: {next_moves[0][1]}')
print(f'Match: {expected_move1 == next_moves[0][0]}')
```

```python
# show frame
sample_frame2
```

```python
# show expected and actual moves
print(f'Expected next move: {expected_move2}')
print(
    f'Predicted next move: {next_moves[1][0]}. Probability: {next_moves[1][1]}')
print(f'Match: {expected_move2 == next_moves[1][0]}')
```
